```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/lena256.jpg')
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
print(gimg.shape)
cv2_imshow(gimg)
```

```python
from random import randint
```

```python
noiseimg = gimg.copy()
noiseRatio = 10;

for h in range(0, gimg.shape[0]):
  for w in range(0, gimg.shape[1]):
    val = randint(0, 100)
    if(val < noiseRatio/2):
      noiseimg[h][w] = 255
    elif(val < noiseRatio):
      noiseimg[h][w] = 0

cv2_imshow(noiseimg)
```

```python
outimg = cv2.medianBlur(noiseimg,5)
cv2_imshow(outimg)
```
